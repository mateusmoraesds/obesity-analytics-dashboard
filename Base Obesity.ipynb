```python
import pandas as pd
import numpy as np
from pandas_gbq import to_gbq
```

```python
colunas = [
    "Gender",
    "Age",
    "Height",
    "Weight",
    "family_history",
    "FAVC",
    "FCVC",
    "NCP",
    "CAEC",
    "SMOKE",
    "CH2O",
    "SCC",
    "FAF",
    "TUE",
    "CALC",
    "MTRANS",
    "Obesity"
]

base = pd.read_csv(
    "gs://tech4-base-obesity/Obesity.csv",
    usecols=colunas,
    encoding="latin1",
    sep=","
)
```

```python
to_gbq(
    base,
    destination_table="tech4_obesity_database.base_obesity_tratada",
    project_id="tech-4-obesity",
    if_exists="replace"
)
```
